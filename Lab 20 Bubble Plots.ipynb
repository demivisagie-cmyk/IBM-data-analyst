### Task 2: Analyzing Relationships Using Bubble Plots


#### 1. Bubble Plot of Technology Preferences by Age

- Visualize the popularity of programming languages respondents have worked with (`LanguageHaveWorkedWith`) across age groups.

- Use bubble size to represent the frequency of each language.



```python
##Write your code here 
import pandas as pd
import matplotlib.pyplot as plt

# Convert age groups to numeric values
age_map = {
    "Under 18 years old": 17,
    "18-24 years old": 21,
    "25-34 years old": 30,
    "35-44 years old": 40,
    "45-54 years old": 50,
    "55-64 years old": 60,
    "65 years or older": 70
}

df["Age_num"] = df["Age"].map(age_map)

plot_df = df[["LanguageHaveWorkedWith", "Age_num"]].dropna()

# Use the first language listed
plot_df["Language"] = plot_df["LanguageHaveWorkedWith"].str.split(";").str[0]

# Bubble size = frequency of the language
freq = plot_df["Language"].value_counts()
plot_df["BubbleSize"] = plot_df["Language"].map(freq)

# Convert language names to numbers for plotting
plot_df["LanguageCode"] = pd.factorize(plot_df["Language"])[0]

plt.figure(figsize=(12,6))

plt.scatter(
    plot_df["Age_num"],
    plot_df["LanguageCode"],
    s=plot_df["BubbleSize"]*2,
    alpha=0.5
)

plt.yticks(
    range(len(plot_df["Language"].unique())),
    plot_df["Language"].unique()
)

plt.xlabel("Age")
plt.ylabel("Programming Language")
plt.title("Technology Preferences by Age")
plt.grid(True)
plt.tight_layout()
plt.show()
```

#### 2. Bubble Plot for Preferred Databases vs. Job Satisfaction

- Explore the relationship between preferred databases (`DatabaseWantToWorkWith`) and job satisfaction.

- Use bubble size to indicate the number of respondents for each database.


```python
##Write your code here
import pandas as pd
import matplotlib.pyplot as plt

# Convert JobSat to numeric
df["JobSat"] = pd.to_numeric(df["JobSat"], errors="coerce")

plot_df = df[["DatabaseWantToWorkWith", "JobSat"]].dropna()

# Use the first preferred database
plot_df["Database"] = plot_df["DatabaseWantToWorkWith"].str.split(";").str[0]

# Bubble size = number of respondents
freq = plot_df["Database"].value_counts()
plot_df["BubbleSize"] = plot_df["Database"].map(freq)

# Convert database names to numbers
plot_df["DatabaseCode"] = pd.factorize(plot_df["Database"])[0]

plt.figure(figsize=(12,6))

plt.scatter(
    plot_df["DatabaseCode"],
    plot_df["JobSat"],
    s=plot_df["BubbleSize"]*2,
    alpha=0.5
)

plt.xticks(
    range(len(plot_df["Database"].unique())),
    plot_df["Database"].unique(),
    rotation=90
)

plt.xlabel("Preferred Database")
plt.ylabel("Job Satisfaction")
plt.title("Preferred Databases vs Job Satisfaction")
plt.grid(True)
plt.tight_layout()
plt.show()
```

### Task 3: Comparing Data Using Bubble Plots


#### 1. Bubble Plot for Compensation Across Developer Roles

- Visualize compensation (`ConvertedCompYearly`) across different developer roles (`DevType`).

- Use bubble size to represent job satisfaction.


```python
##Write your code here
import pandas as pd
import matplotlib.pyplot as plt

df["ConvertedCompYearly"] = pd.to_numeric(df["ConvertedCompYearly"], errors="coerce")
df["JobSat"] = pd.to_numeric(df["JobSat"], errors="coerce")

plot_df = df[["DevType", "ConvertedCompYearly", "JobSat"]].dropna()

# Keep only the first developer role if multiple exist
plot_df["DevType"] = plot_df["DevType"].str.split(";").str[0]

# Convert developer roles to numeric positions
plot_df["DevCode"] = pd.factorize(plot_df["DevType"])[0]

plt.figure(figsize=(12,6))

plt.scatter(
    plot_df["DevCode"],
    plot_df["ConvertedCompYearly"],
    s=plot_df["JobSat"]*20,
    alpha=0.5
)

plt.xticks(
    range(len(plot_df["DevType"].unique())),
    plot_df["DevType"].unique(),
    rotation=90
)

plt.xlabel("Developer Role")
plt.ylabel("ConvertedCompYearly")
plt.title("Compensation Across Developer Roles")
plt.tight_layout()
plt.show()
```

### Task 4: Visualizing Technology Trends Using Bubble Plots


#### 1. Bubble Plot for Preferred Web Frameworks vs. Job Satisfaction

- Explore the relationship between preferred web frameworks (`WebframeWantToWorkWith`) and job satisfaction.

- Use bubble size to represent the number of respondents.



```python
##Write your code here
import pandas as pd
import matplotlib.pyplot as plt

df["JobSat"] = pd.to_numeric(df["JobSat"], errors="coerce")

plot_df = df[["WebframeWantToWorkWith","JobSat"]].dropna()

plot_df["Framework"] = plot_df["WebframeWantToWorkWith"].str.split(";").str[0]

freq = plot_df["Framework"].value_counts()

plot_df["BubbleSize"] = plot_df["Framework"].map(freq)

plot_df["FrameworkCode"] = pd.factorize(plot_df["Framework"])[0]

plt.figure(figsize=(12,6))

plt.scatter(
    plot_df["FrameworkCode"],
    plot_df["JobSat"],
    s=plot_df["BubbleSize"]*2,
    alpha=0.5
)

plt.xticks(
    range(len(plot_df["Framework"].unique())),
    plot_df["Framework"].unique(),
    rotation=90
)

plt.xlabel("Preferred Web Framework")
plt.ylabel("Job Satisfaction")
plt.title("Preferred Web Frameworks vs Job Satisfaction")
plt.tight_layout()
plt.show()
```

#### 2. Bubble Plot for Admired Technologies Across Countries

- Visualize the distribution of admired technologies (`LanguageAdmired`) across different countries (`Country`).

- Use bubble size to represent the frequency of admiration.



```python
##Write your code here
import pandas as pd
import matplotlib.pyplot as plt

plot_df = df[["LanguageAdmired","Country"]].dropna()

plot_df["Language"] = plot_df["LanguageAdmired"].str.split(";").str[0]

counts = (
    plot_df.groupby(["Country","Language"])
    .size()
    .reset_index(name="Count")
)

counts["CountryCode"] = pd.factorize(counts["Country"])[0]
counts["LangCode"] = pd.factorize(counts["Language"])[0]

plt.figure(figsize=(14,7))

plt.scatter(
    counts["CountryCode"],
    counts["LangCode"],
    s=counts["Count"]*2,
    alpha=0.5
)

plt.xticks(
    range(len(counts["Country"].unique())),
    counts["Country"].unique(),
    rotation=90
)

plt.yticks(
    range(len(counts["Language"].unique())),
    counts["Language"].unique()
)

plt.xlabel("Country")
plt.ylabel("Admired Technology")
plt.title("Admired Technologies Across Countries")
plt.tight_layout()
plt.show()
```
