### Step 3: Handle Missing Data


- Identify missing values in the dataset.

- Impute or remove missing values as necessary to ensure data completeness.



```python
## Write your code here
missing_data=df.isnull()
missing_data.head()
df.dropna(axis=0, inplace=True)
df.reset_index(drop=True, inplace=True)
df.head()
```

### Step 4: Analyze Key Columns


- Examine key columns such as `Employment`, `JobSat` (Job Satisfaction), and `YearsCodePro` (Professional Coding Experience).

- **Instruction**: Calculate the value counts for each column to understand the distribution of responses.



```python
## Write your code here
data_url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv'
df = pd.read_csv(data_url)
df["Employment"].value_counts() 
df["JobSat"].value_counts()
df["YearsCodePro"].value_counts()
```

### Step 5: Visualize Job Satisfaction (Focus on JobSat)


- Create a pie chart or KDE plot to visualize the distribution of `JobSat`.

- Provide an interpretation of the plot, highlighting key trends in job satisfaction.


```python
## Write your code here
import matplotlib.pyplot as plt

# Remove missing values
jobsat = df['JobSat'].dropna()

# Create the plot
plt.figure(figsize=(8,5))
jobsat.value_counts().plot(kind='bar')

plt.title('Distribution of Job Satisfaction')
plt.xlabel('Job Satisfaction')
plt.ylabel('Number of Respondents')
plt.xticks(rotation=45)
plt.show()
```

### Step 6: Programming Languages Analysis


- Compare the frequency of programming languages in `LanguageHaveWorkedWith` and `LanguageWantToWorkWith`.
  
- Visualize the overlap or differences using a Venn diagram or a grouped bar chart.


```python
## Write your code here
import pandas as pd
import matplotlib.pyplot as plt

# Count each language mentioned
worked = df['LanguageHaveWorkedWith'].str.get_dummies(sep=';').sum()
want = df['LanguageWantToWorkWith'].str.get_dummies(sep=';').sum()

# Combine the results
languages = pd.DataFrame({
    'Worked With': worked,
    'Want To Work With': want
}).fillna(0)

# Select the top 10 languages by total mentions
top10 = languages.sum(axis=1).sort_values(ascending=False).head(10).index
languages.loc[top10].plot(kind='bar', figsize=(12,6))

plt.title('Top 10 Programming Languages: Worked With vs Want To Work With')
plt.xlabel('Programming Language')
plt.ylabel('Number of Respondents')
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```

### Step 7: Analyze Remote Work Trends


- Visualize the distribution of RemoteWork by region using a grouped bar chart or heatmap.


```python
## Write your code here
import pandas as pd
import matplotlib.pyplot as plt

# Create a cross-tabulation of Country and RemoteWork
remote_region = pd.crosstab(df['Country'], df['RemoteWork'])

# Plot the top 10 countries by number of responses
top10 = remote_region.sum(axis=1).sort_values(ascending=False).head(10).index

remote_region.loc[top10].plot(kind='bar', figsize=(12,6))

plt.title('Remote Work Distribution by Country')
plt.xlabel('Country')
plt.ylabel('Number of Respondents')
plt.xticks(rotation=45)
plt.legend(title='Remote Work')
plt.tight_layout()
plt.show()
```

### Step 8: Correlation between Job Satisfaction and Experience


- Analyze the correlation between overall job satisfaction (`JobSat`) and `YearsCodePro`.
  
- Calculate the Pearson or Spearman correlation coefficient.


```python
## Write your code here
print(df['JobSat'].unique())
print(df['YearsCodePro'].unique())
# Select the required columns
corr_data = df[['JobSat', 'YearsCodePro']].copy()

# Convert to numeric
corr_data['JobSat'] = pd.to_numeric(corr_data['JobSat'], errors='coerce')
corr_data['YearsCodePro'] = pd.to_numeric(corr_data['YearsCodePro'], errors='coerce')

# Remove missing values
corr_data = corr_data.dropna()

# Calculate Spearman correlation
corr = corr_data.corr(method='spearman')

print(corr)
```

### Step 9: Cross-tabulation Analysis (Employment vs. Education Level)


- Analyze the relationship between employment status (`Employment`) and education level (`EdLevel`).

- **Instruction**: Create a cross-tabulation using `pd.crosstab()` and visualize it with a stacked bar plot if possible.


```python
## Write your code here
import pandas as pd
import matplotlib.pyplot as plt

# Create cross-tabulation
cross_tab = pd.crosstab(df['Employment'], df['EdLevel'])

# Plot stacked bar chart
cross_tab.plot(kind='bar', stacked=True, figsize=(12,6))

plt.title('Employment Status vs Education Level')
plt.xlabel('Employment Status')
plt.ylabel('Number of Respondents')
plt.legend(title='Education Level', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
